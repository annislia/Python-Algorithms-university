```python
#5
def RectPS(x1,y1,x2,y2):
    a = abs(x2 - x1)  # длина по оси X
    b = abs(y2 - y1)  # длина по оси Y
    P = 2 * (a + b)
    S = a * b
    return P, S
print(RectPS(1,1,4,5))
```

```python
#5
def MtrProd(a,b):
    r=[]
    
    #размеры изначальных матриц
    c1=len(a[0]) #количество столбцов в 1 матрице
    c2= len(b) #количество строк во 2 матрице
    
    #размеры итоговой матрицы
    r1= len(a)
    r2=len(b[0])
    
    #создаем пустую матрицу нужного размера заполненную 0
    for i in range(r1):
        ns=[]
        for j in range(r2):
            ns.append(0)
        r.append(ns)

    #произведение
    for i in range(r1):
        for j in range(r2):
            s=0
            for k in range(c1):
                s+= a[i][k]*b[k][j]
            r[i][j]=s
    return r
    



mA = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]]

mB = [
    [9, 8, 7],
    [6, 5, 4],
    [3, 2, 1]
]
print(MtrProd(mA,mB))
```

```python
#6
def SumMtr(a,b):
    r=[]
    for i in range(len(a)):
        ns=[]
        for j in range(len(a[i])):
            s=a[i][j]+b[i][j]
            ns.append(s)
        r.append(ns)
    return r
            

    
mA = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]]

mB = [
    [9, 8, 7],
    [6, 5, 4],
    [3, 2, 1]
]
print(SumMtr(mA,mB))
```

```python
#9
def InvertStr(k, k1, k2):
    part1= k[:k1]
    in1=min(k1+k2, len(k))
    parttoinvert= k[k1: in1]
    part2=k[k2:]
    inv= parttoinvert[::-1]
    r= part1+inv+part2
    return r

print(InvertStr("abcdefghijk",2,4))
```

```python
#11
def CompressStr(s):
    for i in s:
        k=s.count(i)
        if k>3:
            for j in range(k):
                if i*(k-j) in s:
                    s=s.replace(i*k, i+'{'+str(k)+'}')
                    break
    return s
                
                
s='bbbccccce'
print(CompressStr(s))
```

```python
#13
def Chessboard(m,n):
    a=[]
    for i in range(m):
        ns=[]
        for j in range(n):
            v=(i+j)%2
            ns.append(v)
        a.append(ns)
   
    return a
            
            

print(Chessboard(3,3))
```

```python
#14
def Bell(l):
    r=[]
    l=sorted(l)
    while l:
        r.append(l.pop(0))
        if l:
            r.append(l.pop())
    return r
l = [3, 1, 4, 1, 5, 9, 2, 6]
print(Bell(l))
```
